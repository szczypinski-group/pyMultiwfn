## Note
It is important to note that some Multiwfn calculations require specific file types, or cannot process specific file types. These will be specified for the relevant examples in the tutorial but for further calculations, the user should refer to the [Multiwfn manual](http://sobereva.com/multiwfn/misc/Multiwfn_manual_2026.1.12.pdf). This will be implemented in future version to allow a clearer debugging route for users.

## Charges and Population Calculations
Multiwfn allows the user to calcualte a variety of different charges. In this instance the benzene.wfn file is used to calcualte the Hirhsfeld charges, as such. The run() function needs to be called to run the calculation. For this set of calculations, the standard molden/wfn/wfx files shoul

```python
from pathlib import Path
from pymultiwfn import MultiwfnAnalysis, Multiwfn, Menu

analysis = MultiwfnAnalysis("benzene.wfn", analyses=[Menu.HIRSHFELD_CHARGE])
analysis.run()

result = analysis.results[0]          # MultiwfnResult for HIRSHFELD_CHARGE
for item in result.result:
    print(item)                        
```

Multiple calculations can also be done in the same input. In this case, we will be calculating both the Becke charges and CHELPG charges. This follows the same premise as the previous calculation.

```python
from pymultiwfn import MultiwfnAnalysis, Menu

multiple = MultiwfnAnalysis("benzene.wfn", analyses=[
    Menu.BECKE_CHARGE,
    Menu.CHELPG_CHARGE,
])

multiple.run()
print(multiple.results[0].to_dict())
```

Finally, we can also run calculations for an entire family, which are sorted based on their menu category. Charges, are started from the "7" input. The results can also be printed to a specified output file. This is specified when the run function is called.

```python
from pymultiwfn import MultiwfnAnalysis, AnalysisClass, Menu
from pymultiwfn.enums import analyses
from pathlib import Path

all = MultiwfnAnalysis("benzene.wfn", analyses=[
    analyses.CHARGES
])

all.run(work_dir=Path("results"))
print(all.results[0].to_dict())
```

## Spectra Plotting
Multiwfn is able to plot a variety of spectra. The required input can vary for each spectra, but will be specified.

### IR Spectra Plotting
The process for plotting an IR spectra is fairly straightforward. The idea remains the same as the previously described calculations. The only difference is the required file type. the standard wfn/molden file does not have the required information to plot an IR spectra, a Guassian or ORCA output must be used to plot this spectra. The output should be 

```python
from pymultiwfn import MultiwfnAnalysis, Menu


ir = MultiwfnAnalysis("NH3BF3_freq.out", analyses=[
    Menu.PLOT_IR_SPECTRUM
])

ir.run()
print(ir.results[0].to_dict())
```

### UV-Vis
Once again, the process to plot a spectra remains the same. Simnilarly to IR, the required input must be either a Gaussian or ORCA output file. 


```python
from pymultiwfn import MultiwfnAnalysis, Menu

uv = MultiwfnAnalysis("NH3BF3_freq.out", analyses=[
    Menu.PLOT_UV_VIS_SPECTRUM
])

uv.run()
print(uv.results[0].to_dict())
```
